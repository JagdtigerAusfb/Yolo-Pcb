Bloco 1: Importações e Configurações Iniciais

Rode este bloco sempre que abrir o arquivo para carregar as bibliotecas na memória.

```python
import os
import zipfile
from ultralytics import YOLO

# Configurações do seu TCC
ARQUIVO_ZIP = "data.zip"
PASTA_DESTINO = "dataset_extraido"
NOME_DAS_CLASSES = [
    'mouse_bite',
    'spur',
    'missing_hole',
    'short',
    'open_circuit',
    'spurious_copper'
]

print("done")
```

Bloco 2: Extração do Dataset

Rode este bloco apenas uma vez. Ele é inteligente o suficiente para pular a extração se a pasta já existir.

```python
if not os.path.exists(PASTA_DESTINO):
    print(f"📦 Extraindo '{ARQUIVO_ZIP}'...")
    with zipfile.ZipFile(ARQUIVO_ZIP, 'r') as zip_ref:
        zip_ref.extractall(PASTA_DESTINO)
    print(f"✅ Arquivos extraídos com sucesso para a pasta: {PASTA_DESTINO}")
else:
    print(f"📂 Pasta '{PASTA_DESTINO}' já existe. Pulando extração.")
```

Bloco 3: Criação do Arquivo de Configuração (YAML)

Este bloco gera o arquivo de texto para o YOLO.

```python
caminho_base = os.path.abspath("/content/dataset_extraido/data")
caminho_yaml = os.path.join(caminho_base, "data.yaml")

yaml_content = f"""
path: {caminho_base}
train: train/images
val: val/images

nc: {len(NOME_DAS_CLASSES)}
names: {NOME_DAS_CLASSES}
"""

with open(caminho_yaml, 'w', encoding='utf-8') as f:
    f.write(yaml_content.strip())
print(f"📝 Arquivo 'data.yaml' atualizado em: {caminho_yaml}")
```

Bloco 4: O Motor de IA

Este é o bloco pesado. É aqui que você vai mexer nas épocas (epochs) e no tamanho do lote (batch) para testar a performance do seu modelo.

```python
print("🚀 Carregando o modelo YOLOv8 Nano...")

# ==============================================================================
# OPÇÕES DE MODELOS YOLO (Basta trocar o nome do arquivo .pt abaixo)
# ==============================================================================
# YOLOv8 para Detecção de Objetos (Do mais leve/rápido ao mais pesado/preciso):
#
# 'yolov8n.pt' -> Nano  (Mais rápido, ideal para testes e hardware limitado)
# 'yolov8s.pt' -> Small (Ótimo equilíbrio entre velocidade e precisão)
# 'yolov8m.pt' -> Medium(Boa precisão para a maioria dos projetos de TCC)
# 'yolov8l.pt' -> Large (Alta precisão, mas exige bastante da GPU)
# 'yolov8x.pt' -> XLarge(Máxima precisão, treinamento mais demorado)
#
# Dica: A biblioteca ultralytics também suporta modelos mais recentes,
# bastando usar 'yolov9c.pt', 'yolov10n.pt' ou 'yolo11n.pt', por exemplo.
# ==============================================================================

model = YOLO('yolov8m.pt') # <- Altere aqui se quiser testar os modelos maiores

print("🔥 Iniciando o treinamento...")
results = model.train(
    data=caminho_yaml,
    epochs=60,
    imgsz=640,
    batch=16,
    name='tcc_pcb_colab',
    device=0, # O zero indica o uso da GPU NVIDIA do Colab
    plots=True
)

print("\n✅ Treinamento finalizado!")
```
